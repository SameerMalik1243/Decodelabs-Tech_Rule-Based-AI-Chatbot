## Rule-Based AI Chatbot🤖


#### As an Artificial Intelligence Intern at DecodeLabs, I am engineering logic-driven foundational systems.
#### This notebook demonstrates a "White Box" rule-based chatbot built entirely on Python's control flow mechanics.
#### The resulting architecture guarantees predictable, accurate responses without the risk of AI hallucinations.

### The Rules Dictionary

#### Create a dictionary to map specific user inputs to bot responses.

```python
bot_rules = {
    # Greetings
    'hello': "Hello there! How can I help you with your industrial training today?",
    'hi': "Hello there! How can I help you with your industrial training today?",
    'hey': "Hello there! How can I help you with your industrial training today?",
    'greetings': "Hello there! How can I help you with your industrial training today?",
    
    # Project Info
    'what is this project': "This is Project 1: The Rule-Based AI Chatbot. It teaches deterministic AI concepts.",
    'project': "This is Project 1: The Rule-Based AI Chatbot. It teaches deterministic AI concepts.",
    'project 1': "This is Project 1: The Rule-Based AI Chatbot. It teaches deterministic AI concepts.",
    
    # DecodeLabs Info
    'what is decodelabs': "DecodeLabs is an educational platform providing hands-on industrial training.",
    'decodelabs': "DecodeLabs is an educational platform providing hands-on industrial training."
}

# The default message if the user says something not in the dictionary
fallback_msg = "I'm sorry, I only understand specific commands. Try saying 'hi', asking about 'Project', or typing 'exit'."
```

### The Chatbot Logic
#### This implementation constructs a deterministic "White Box" architecture by utilizing string normalization for input 
#### sanitization and a dictionary-based logic engine to ensure traceable, zero-hallucination responses.  

```python
def run_chatbot():
    print("Welcome to the DecodeLabs Rule-Based Chatbot!")
    print("Ask me about 'Project 1' or 'DecodeLabs'. (Type 'exit' to quit)")
    
    while True:
        # 1. INPUT
        raw_input = input("\nYou: ")
        
        # 2. SANITIZATION
        clean_input = raw_input.lower().strip()
        
        if clean_input in ['bye', 'quit', 'exit']:
            print("Chatbot: Goodbye! Best of luck with your DecodeLabs training!")
            break
            
        # 3. PROCESS: Using the .get() method.
        # It tries to find the 'clean_input' in bot_rules. 
        # If it doesn't exist, it automatically returns 'fallback_msg'.
        bot_response = bot_rules.get(clean_input, fallback_msg)
            
        # 4. OUTPUT
        print(f"Chatbot: {bot_response}")
```

###  Run the Program

```python
# Start the continuous loop
run_chatbot()
```
